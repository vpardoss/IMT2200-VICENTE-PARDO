### 3.1 DataFrame unificado (1 punto)

Vamos a cargar en un DataFrame los datos de juegos guardados en el documento `bgg_GameItem.csv`. Al inspeccionar nuestros datos, podemos notar que los valores de ciertas comunas vienen "codificados" con ID. Para comenzar, vamos a juntar la información de los distintos archivos descargados.

**a) (0.8 pts)** Genere un DataFrame único con toda la información de cada juego, incluyendo: nombres de artistas y diseñadores involucrados, mecánicas, categorías, tipo, familia y editorial. Guarde este DataFrame en un nuevo archivo CSV.

```python
# Cargar los datos
gameitems = pd.read_csv("Data/board-games/bgg_GameItem.csv")
categories = pd.read_csv("Data/board-games/bgg_Category.csv")
gamefamilies = pd.read_csv("Data/board-games/bgg_GameFamily.csv")
gametypes = pd.read_csv("Data/board-games/bgg_GameType.csv")
mechanics = pd.read_csv("Data/board-games/bgg_Mechanic.csv")
people = pd.read_csv("Data/board-games/bgg_Person.csv")
publishers = pd.read_csv("Data/board-games/bgg_Publisher.csv")

'''
Al no saber como unificar los DataFrames algunos venían con un bgg_id y otros no, consulté a una
IA sobre como podía hacerlo utilizando el siguiente prompt:

"Necesito unificar los siguientes DataFrames de pandas en un solo DataFrame unificado.
Mi problema es que algunos dataframes contienen una columna de "bdd_id" y otros no. Como
podría unificar unificar estos DataFrames?

Los dataframes se ven así:

gameitems = pd.read_csv("Data/board-games/bgg_GameItem.csv")
categories = pd.read_csv("Data/board-games/bgg_Category.csv")
gamefamilies = pd.read_csv("Data/board-games/bgg_GameFamily.csv")
gametypes = pd.read_csv("Data/board-games/bgg_GameType.csv")
mechanics = pd.read_csv("Data/board-games/bgg_Mechanic.csv")
people = pd.read_csv("Data/board-games/bgg_Person.csv")
publishers = pd.read_csv("Data/board-games/bgg_Publisher.csv")
"
A lo cual la IA (Claude) mencionó que conocía la estructura de BoardGameGeek, y creó la siguiente función:
'''
# Función para convertir IDs separados por comas a nombres
def map_ids_to_names(id_string, mapping_dict):
    if pd.isna(id_string) or id_string == '':
        return ''
    
    ids = str(id_string).split(',')
    names = []
    for id_str in ids:
        try:
            id_num = int(id_str.strip())
            if id_num in mapping_dict:
                names.append(mapping_dict[id_num])
        except ValueError:
            continue
    
    return ', '.join(names)

# Fin de código de IA

'''
Mencionando que:

"La función map_ids_to_names():

Toma una cadena de IDs separados por comas (ej: "133,2,24883")
Los divide y convierte a enteros
Busca cada ID en el diccionario correspondiente
Devuelve los nombres separados por comas"

También recomendó crear diccionarios de mapeo para cada DataFrame.
'''
# Creamos diccionarios de mapeo para cada DataFrame
categories_dict = categories.set_index("bgg_id")["name"].to_dict()
mechanics_dict = mechanics.set_index("bgg_id")["name"].to_dict()
gametypes_dict = gametypes.set_index("bgg_id")["name"].to_dict()
gamefamilies_dict = gamefamilies.set_index("bgg_id")["name"].to_dict()
people_dict = people.set_index("bgg_id")["name"].to_dict()
publishers_dict = publishers.set_index("bgg_id")["name"].to_dict()

# Creamos el DataFrame unificado copiando gameitems
df_unificado = gameitems.copy()

'''
Ahora debía aplicar el mapeo a cada columna, por lo que le pregunté nuevamente a la IA
cómo podía hacerlo usando el siguiente prompt:
"Logré crear los diccionarios de mapeo para cada dataframe con la documentación.
Ahora, como puedo aplicar el mapeo a la copia de gameitems? Gracias."

Y creó el siguiente código.
'''
# Aplicar el mapeo a cada columna relevante
df_unificado['game_type_names'] = df_unificado['game_type'].apply(
    lambda x: map_ids_to_names(x, gametypes_dict)
)

df_unificado['designer_names'] = df_unificado['designer'].apply(
    lambda x: map_ids_to_names(x, people_dict)
)

df_unificado['artist_names'] = df_unificado['artist'].apply(
    lambda x: map_ids_to_names(x, people_dict)
)

df_unificado['publisher_names'] = df_unificado['publisher'].apply(
    lambda x: map_ids_to_names(x, publishers_dict)
)

df_unificado['category_names'] = df_unificado['category'].apply(
    lambda x: map_ids_to_names(x, categories_dict)
)

df_unificado['mechanic_names'] = df_unificado['mechanic'].apply(
    lambda x: map_ids_to_names(x, mechanics_dict)
)

df_unificado['family_names'] = df_unificado['family'].apply(
    lambda x: map_ids_to_names(x, gamefamilies_dict)
)
# Fin de código de IA

# Por último guardamos el DataFrame completo en un nuevo archivo .csv
df_unificado.to_csv("Data/df_unificado.csv", index=False)
```

**a) (0.3 pts)** Inspecciones y filtre los datos que tengan valores válidos para: tiempo mínimo de juego, tiempo máximo de juego y complejidad. En el caso de que existan "outliers", puede descartarlos, cosiderando un rango razonable para las variables anteriores. Justifique su desición.

```python
print(f"Cantidad de registros original: {len(df_unificado)}")

# Eliminamos valores nulos
df_filtrado = df_unificado.dropna(subset=['min_time', 'max_time', 'complexity']).copy()
'''
Prompt de IA: "Quiero saber como puedo filtrar los datos de duración de tiempo (min_time, max_time, complexity)
para eliminar outliers en un rango razonable. limítate a enseñarme como funcionaría la función de filtrado.
'''
df_filtrado = df_filtrado[(df_filtrado['min_time'] >= 1) & (df_filtrado['min_time'] <= 300) &
    (df_filtrado['max_time'] >= 1) & (df_filtrado['max_time'] <= 720)
].copy()
# Fin de código de IA

# Bajo esa misma lógica, eliminaré los valores donde max_time es menor a min_time porque no tienen sentido.
df_filtrado = df_filtrado[df_filtrado["max_time"] >= df_filtrado["min_time"]].copy()
print(f"Cantidad despues de aplicar filtros: {len(df_filtrado)}")
```

Justificación: En min_time, el 50% corresponde a 30 minutos y el 75% a 60 minutos, un máximo de 300 minutos es más que prudente. En max_time, 12 horas debería ser suficiente para cualquier juego de mesa, y en complexity el valor máximo, 5, está bien por lo que no necesita filtro.

BoardGameGeek permite evaluar la complejidad (o "weight") de un juego en una escala de 1 a 5, siendo 1 un juego "liviano" o fácil de entender, y 5 un juego "pesado" o complejo. Primero, seleccione los juegos que han sido evaluados por al menos 100 usuarios. Luego grafique la complejidad promedio de los juegos según año. Responda: ¿ha cambiado la percepción de complejidad entre juegos entre 1980 y la actualidad?

```python
print(f"Total de juegos antes del filtro: {len(df_filtrado)}")

# Filtramos votos nulos
votos_nulos = df_filtrado["num_votes"].isnull().sum()
print(f"Despues del filtro: {len(df_filtrado) - votos_nulos}")

juegos_evaluados = df_filtrado[(df_filtrado["num_votes"] >= 100) & (df_filtrado["complexity"].notnull()) & (df_filtrado["year"].between(1980, 2025))].copy()
print(f"Juegos que poseen más de 100 votos: {len(juegos_evaluados)}")

'''
Prompt de IA: "Necesito calcular la complejidad promedio por año para después graficarla, de ser posible,
necesito que sea un gráfico completo.

votos_nulos = df_filtrado["num_votes"].isnull().sum()
print(f"Despues del filtro: {len(df_filtrado) - votos_nulos}")

juegos_evaluados = df_filtrado[(df_filtrado["num_votes"] >= 100) & (df_filtrado["complexity"].notnull()) & (df_filtrado["year"].between(1980, 2025))].copy()
print(f"Juegos que poseen más de 100 votos: {len(juegos_evaluados)}")
"
'''
complejidad_por_ano = juegos_evaluados.groupby('year')['complexity'].agg(['mean', 'count', 'std']).reset_index()
complejidad_por_ano.columns = ['year', 'complexity_mean', 'num_games', 'complexity_std']

# Crear el gráfico principal
plt.figure(figsize=(16, 12))

# Gráfico de línea con complejidad promedio
plt.subplot(2, 1, 1)
plt.plot(complejidad_por_ano['year'], complejidad_por_ano['complexity_mean'], 
         linewidth=3, color='darkred', marker='o', markersize=5, alpha=0.8)

# Agregar línea de tendencia
z = np.polyfit(complejidad_por_ano['year'], complejidad_por_ano['complexity_mean'], 1)
p = np.poly1d(z)
plt.plot(complejidad_por_ano['year'], p(complejidad_por_ano['year']), 
         "r--", alpha=0.6, linewidth=2, label=f'Tendencia (pendiente: {z[0]:.4f})')

# Marcar el año 2000
plt.axvline(x=2000, color='blue', linestyle='--', alpha=0.7, label='Año 2000')
plt.title('Complejidad Promedio de Juegos por Año (1980-2025)\n(Juegos con ≥100 evaluaciones)', 
          fontsize=16, fontweight='bold')
plt.xlabel('Año', fontsize=12)
plt.ylabel('Complejidad Promedio (1-5)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend()
plt.ylim(1, 5)
# Fin de código de IA
```
